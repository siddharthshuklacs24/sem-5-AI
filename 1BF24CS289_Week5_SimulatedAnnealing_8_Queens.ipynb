```python

import random
import math

# Calculate the number of attacking pairs of queens
def calculate_cost(board):
    conflicts = 0
    n = len(board)

    for i in range(n):
        for j in range(i + 1, n):
            # Check diagonal attacks
            if abs(board[i] - board[j]) == abs(i - j):
                conflicts += 1

    return conflicts


# Generate a neighboring state
def generate_neighbor(board):
    new_board = board.copy()

    col = random.randint(0, 7)
    new_row = random.randint(0, 7)

    while new_row == new_board[col]:
        new_row = random.randint(0, 7)

    new_board[col] = new_row
    return new_board


# Simulated Annealing Algorithm
def simulated_annealing():
    n = 8

    # Random initial state
    board = [random.randint(0, 7) for _ in range(n)]

    temperature = 1000
    cooling_rate = 0.99
    min_temperature = 0.0001

    current_cost = calculate_cost(board)

    while temperature > min_temperature:

        if current_cost == 0:
            break

        # Generate a neighboring solution
        new_board = generate_neighbor(board)
        new_cost = calculate_cost(new_board)

        delta = new_cost - current_cost

        # Accept better solutions
        if delta < 0:
            board = new_board
            current_cost = new_cost

        # Accept worse solutions probabilistically
        elif random.random() < math.exp(-delta / temperature):
            board = new_board
            current_cost = new_cost

        # Reduce temperature
        temperature *= cooling_rate

    return board, current_cost


# Display the chessboard
def display_board(board):
    print("\nChessboard:")

    for row in range(8):
        for col in range(8):
            if board[col] == row:
                print(" Q ", end="")
            else:
                print(" . ", end="")
        print()


# Main program
if __name__ == "__main__":

    solution, cost = simulated_annealing()

    print("8 Queens Problem using Simulated Annealing")
    print("Final Cost:", cost)

    if cost == 0:
        print("Solution Found!")
        print("Queen Positions:", solution)
    else:
        print("Solution not found. Try running again.")

    display_board(solution)
```
