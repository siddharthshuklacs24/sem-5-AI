```python

# FOL Forward Chaining

# Facts in the knowledge base
facts = [
    ("Human", "Socrates")
]

# Rules: (premises, conclusion)
rules = [
    ([("Human", "?x")], ("Mortal", "?x"))
]


# Function to match a fact with a rule premise
def match(pattern, fact, substitution):
    if pattern[0] != fact[0]:
        return None

    new_substitution = substitution.copy()

    for p, f in zip(pattern[1:], fact[1:]):
        if p.startswith("?"):
            if p in new_substitution:
                if new_substitution[p] != f:
                    return None
            else:
                new_substitution[p] = f
        elif p != f:
            return None

    return new_substitution


# Apply substitutions to generate a new fact
def apply_rule(conclusion, substitution):
    return tuple(
        substitution.get(item, item)
        for item in conclusion
    )


# Forward chaining algorithm
def forward_chaining(facts, rules, query):

    facts = list(facts)

    while True:
        new_facts = []

        for premises, conclusion in rules:

            # Start with an empty substitution
            substitutions = [{}]

            # Match every premise with existing facts
            for premise in premises:
                updated = []

                for substitution in substitutions:
                    for fact in facts:
                        result = match(
                            premise, fact, substitution
                        )

                        if result is not None:
                            updated.append(result)

                substitutions = updated

            # Generate conclusions
            for substitution in substitutions:
                new_fact = apply_rule(
                    conclusion, substitution
                )

                if new_fact not in facts:
                    new_facts.append(new_fact)

        # If no new facts are generated, stop
        if not new_facts:
            break

        facts.extend(new_facts)

        for fact in new_facts:
            print("New fact:", fact)

        if query in facts:
            return True

    return query in facts


# Query
query = ("Mortal", "Socrates")

if forward_chaining(facts, rules, query):
    print("Query proved:", query)
else:
    print("Query cannot be proved")
```
