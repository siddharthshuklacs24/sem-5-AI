```python
from itertools import product

def main():

    print("P\tQ\tR\tKB\tQuery")

    for values in product([True, False], repeat=3):
        p = values[0]
        q = values[1]
        r = values[2]

        kb = ((not p) or q) and p and ((not q) or r)

        query = r

        print(p, "\t", q, "\t", r, "\t", kb, "\t", query)

        if kb and not query:
            print("\nKB does not entail the Query.")
            return

    print("\nKB entails the Query.")


main()
```
