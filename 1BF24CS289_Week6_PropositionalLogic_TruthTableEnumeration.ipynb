```python

from itertools import product

# Evaluate the knowledge base
def KB(P, Q):
    return P and ((not P) or Q)

# Evaluate the query
def query(P, Q):
    return Q

# Truth table enumeration
def entails():
    symbols = ["P", "Q"]
    result = True

    print("P\tQ\tKB\tQuery")

    for values in product([True, False], repeat=len(symbols)):
        P, Q = values

        kb_value = KB(P, Q)
        query_value = query(P, Q)

        print(P, "\t", Q, "\t", kb_value, "\t", query_value)

        # Check whether KB is true but query is false
        if kb_value and not query_value:
            result = False

    if result:
        print("\nKB entails the Query.")
    else:
        print("\nKB does NOT entail the Query.")


entails()
```
