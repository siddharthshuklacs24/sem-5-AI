```python

# Check whether a term is a variable
def is_variable(x):
    return isinstance(x, str) and x[0].islower()


# Apply substitutions to a term
def substitute(term, subs):
    if isinstance(term, str):
        while term in subs:
            term = subs[term]
        return term

    return (term[0], tuple(substitute(arg, subs)
                           for arg in term[1]))


# Occurs check
def occurs(var, term, subs):
    term = substitute(term, subs)

    if var == term:
        return True

    if isinstance(term, tuple):
        return any(occurs(var, arg, subs) for arg in term[1])

    return False


# Unification algorithm
def unify(x, y, subs=None):

    if subs is None:
        subs = {}

    x = substitute(x, subs)
    y = substitute(y, subs)

    # Both expressions are identical
    if x == y:
        return subs

    # If x is a variable
    if is_variable(x):
        if occurs(x, y, subs):
            return None

        subs[x] = y
        return subs

    # If y is a variable
    if is_variable(y):
        if occurs(y, x, subs):
            return None

        subs[y] = x
        return subs

    # Both are compound expressions
    if isinstance(x, tuple) and isinstance(y, tuple):

        # Check predicate/function names
        if x[0] != y[0]:
            return None

        # Check number of arguments
        if len(x[1]) != len(y[1]):
            return None

        # Unify corresponding arguments
        for a, b in zip(x[1], y[1]):
            subs = unify(a, b, subs)

            if subs is None:
                return None

        return subs

    # Expressions cannot be unified
    return None


# Main program
print("First Order Logic - Unification")

# Example: P(x, a) and P(b, a)
expr1 = ("P", ("x", "a"))
expr2 = ("P", ("b", "a"))

result = unify(expr1, expr2)

if result is not None:
    print("Unification Successful")
    print("Substitution:", result)
else:
    print("Unification Failed")
```
