```python

# Week-4 Lab
# Hill Climbing Search Algorithm for N-Queens Problem

N = 4

# Calculate the number of attacking queen pairs
def calculate_cost(state):
    cost = 0

    for i in range(N):
        for j in range(i + 1, N):
            # Check same row or diagonal attack
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                cost += 1

    return cost


# Display chessboard
def display_board(state):
    for row in range(N):
        for col in range(N):
            if state[col] == row:
                print(" Q ", end="")
            else:
                print(" . ", end="")
        print()
    print()


# Generate all neighboring states
def generate_neighbors(state):
    neighbors = []

    for col in range(N):
        for row in range(N):
            if state[col] != row:
                new_state = state.copy()
                new_state[col] = row
                neighbors.append(new_state)

    return neighbors


# Hill Climbing Algorithm
def hill_climbing(initial_state):
    current_state = initial_state

    print("Initial State:")
    print("Configuration:", current_state)
    print("Cost:", calculate_cost(current_state))
    display_board(current_state)

    step = 1

    while True:
        current_cost = calculate_cost(current_state)

        if current_cost == 0:
            print("Goal State Reached!")
            print("Final Configuration:", current_state)
            print("Final Cost:", current_cost)
            display_board(current_state)
            return

        neighbors = generate_neighbors(current_state)

        best_neighbor = None
        best_cost = current_cost

        print("Step:", step)
        print("Current State:", current_state)
        print("Current Cost:", current_cost)

        print("\nNeighbor States and Costs:")

        for neighbor in neighbors:
            cost = calculate_cost(neighbor)

            print(neighbor, "-> Cost:", cost)

            if cost < best_cost:
                best_cost = cost
                best_neighbor = neighbor

        # Check whether a better neighbor exists
        if best_neighbor is None:
            print("\nLocal Minimum Reached!")
            print("No better neighbor found.")
            print("Final State:", current_state)
            print("Final Cost:", current_cost)
            return

        current_state = best_neighbor

        print("\nSelected Best Neighbor:", current_state)
        print("New Cost:", best_cost)
        display_board(current_state)

        step += 1


# Main Program
print("HILL CLIMBING - 4 QUEENS PROBLEM")
print("--------------------------------")

# Initial configuration
initial_state = [1, 3, 0, 2]

hill_climbing(initial_state)
```
