```python
import heapq


# Goal state
GOAL = (
    1, 2, 3,
    4, 5, 6,
    7, 8, 0
)


# Heuristic: Manhattan Distance
def manhattan_distance(state):
    distance = 0

    for i in range(9):

        # Do not calculate distance for blank
        if state[i] == 0:
            continue

        # Current position
        current_row = i // 3
        current_col = i % 3

        # Goal position of this tile
        goal_index = GOAL.index(state[i])

        goal_row = goal_index // 3
        goal_col = goal_index % 3

        # Manhattan distance
        distance += abs(current_row - goal_row)
        distance += abs(current_col - goal_col)

    return distance


# Generate all valid moves
def get_neighbors(state):
    neighbors = []

    zero_index = state.index(0)

    row = zero_index // 3
    col = zero_index % 3

    # Up, Down, Left, Right
    moves = [
        (-1, 0),
        (1, 0),
        (0, -1),
        (0, 1)
    ]

    for dr, dc in moves:

        new_row = row + dr
        new_col = col + dc

        # Check whether the move is inside the board
        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_index = new_row * 3 + new_col

            new_state = list(state)

            # Move the blank
            new_state[zero_index], new_state[new_index] = \
                new_state[new_index], new_state[zero_index]

            neighbors.append(tuple(new_state))

    return neighbors


# Display puzzle state
def print_state(state):
    for i in range(0, 9, 3):
        print(state[i:i + 3])

    print()


# A* Search
def a_star(start):

    # (f, g, state, path)
    open_list = []

    g = 0
    h = manhattan_distance(start)
    f = g + h

    heapq.heappush(
        open_list,
        (f, g, start, [start])
    )

    closed = set()

    while open_list:

        f, g, current, path = heapq.heappop(open_list)

        # Skip already explored states
        if current in closed:
            continue

        closed.add(current)

        # Goal reached
        if current == GOAL:
            return path

        # Generate successor states
        for neighbor in get_neighbors(current):

            if neighbor in closed:
                continue

            new_g = g + 1
            new_h = manhattan_distance(neighbor)
            new_f = new_g + new_h

            heapq.heappush(
                open_list,
                (
                    new_f,
                    new_g,
                    neighbor,
                    path + [neighbor]
                )
            )

    return None


# Initial state
start = (
    1, 2, 3,
    0, 4, 6,
    7, 5, 8
)


# Run A*
solution = a_star(start)


# Display result
if solution:

    print("Solution found!")
    print("Number of moves:", len(solution) - 1)
    print()

    for step, state in enumerate(solution):

        print("Step:", step)
        print_state(state)

else:

    print("No solution found.")
```
