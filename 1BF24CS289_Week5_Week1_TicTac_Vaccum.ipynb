```python
#WEEK 1:TIC TAC TOE IMPLEMENTATION
import random

board=[" "]*9
def print_board():
    print(board[0],"|",board[1],"|",board[2])
    print()
    print(board[3],"|",board[4],"|",board[5])
    print()
    print(board[6],"|",board[7],"|",board[8])
    print()

def board_full():
  return " " not in board

winning_combinations=[
    #first three are rows,next three columns last 2 are diagonals
    [0,1,2],[3,4,5],[6,7,8],[0,3,6],[1,4,7],[2,5,8],[0,4,8],[2,4,6]
]

def human_move():
    while(True):
        position=int(input("enter position(1 to 9):"))

        #position invalid check
        if(position<1 or position>9):
          print("enter valid position")
        #position full  check
        elif(board[position-1]!=" "):
          print("position already filled")
        #correct position entered
        else:
          board[position-1]="X"
          break

def computer_move():
    while(True):
        position=random.randint(0,8)
        if(board[position]==" "):
            board[position]="O"
            break

def check_win(Player):
    for combination in winning_combinations:
        if board[combination[0]]==board[combination[1]]==board[combination[2]]==Player:
            return True
    return False

#main code(the helper functions will be above this)
print("TIC TAC TOE game\n")
print("HUMAN:X\n")
print("COMPUTER:O\n")

while(True):
    #display the board before letting human play
    print_board()

    #human plays
    human_move()

    #display the board after human has played
    print_board()

    #check if human won
    if(check_win("X")):
        print("HUMAN WINS")
        break

    #check if draw happened
    if(board_full()):
        print("DRAW")
        break

    #let the computer play if none of the above 2 conditions happened
    computer_move()

    #check if computer won
    if(check_win("O")):
        print("COMPUTER WINS")
        break

    #check if draw happened
    if(board_full()):
        print("DRAW")
        break


```

```python
#WEEK 1 VACUUM CLEANER IMPLEMENTATION
room_A=input("enter room A's state(dirty/clean):").lower()
room_B=input("enter room B's state(dirty/clean):").lower()
position=input("enter the position of vacuum cleaner(A or B):").upper()

#displaying initial state
print("initial state")
print("room A:",room_A)
print("room B:",room_B)
print("position of vacuum cleaner:",position)
print()

while(True):
  #goal achieved
  if(room_A=="clean" and room_B=="clean"):
    print("goal achieved")
    break

  #vacuum cleaner is in room A
  elif(position=="A"):
    #room A is dirty
    if(room_A=="dirty"):
      print("room A is dirty,cleaning it")
      room_A="clean"
      print("room A:",room_A)
      print("room B:",room_B)
      print("position of vacuum cleaner:",position)
      print()

    #room A is clean
    else:
      print("room A is clean,moving to room B")
      position="B"

    #vacuum cleaner is in room B
  elif(position=="B"):
    #room B is dirty
    if(room_B=="dirty"):
      print("room B is dirty,cleaning it")
      room_B="clean"
      print("room A:",room_A)
      print("room B:",room_B)
      print("position of vacuum cleaner:",position)
      print()

    #room B is clean
    else:
      print("room B is clean,moving to room A")
      position="A"
```
