```python
#WEEK 1:TIC TAC TOE IMPLEMENTATION
import random

board=[" "]*9
def print_board():
    print(board[0],"|",board[1],"|",board[2])
    print()
    print(board[3],"|",board[4],"|",board[5])
    print()
    print(board[6],"|",board[7],"|",board[8])
    print()

def board_full():
  return " " not in board

winning_combinations=[
    #first three are rows,next three columns last 2 are diagonals
    [0,1,2],[3,4,5],[6,7,8],[0,3,6],[1,4,7],[2,5,8],[0,4,8],[2,4,6]
]

def human_move():
    while(True):
        position=int(input("enter position(1 to 9):"))

        #position invalid check
        if(position<1 or position>9):
          print("enter valid position")
        #position full  check
        elif(board[position-1]!=" "):
          print("position already filled")
        #correct position entered
        else:
          board[position-1]="X"
          break

def computer_move():
    while(True):
        position=random.randint(0,8)
        if(board[position]==" "):
            board[position]="O"
            break

def check_win(Player):
    for combination in winning_combinations:
        if board[combination[0]]==board[combination[1]]==board[combination[2]]==Player:
            return True
    return False

#main code(the helper functions will be above this)
print("TIC TAC TOE game\n")
print("HUMAN:X\n")
print("COMPUTER:O\n")

while(True):
    #display the board before letting human play
    print_board()

    #human plays
    human_move()

    #display the board after human has played
    print_board()

    #check if human won
    if(check_win("X")):
        print("HUMAN WINS")
        break

    #check if draw happened
    if(board_full()):
        print("DRAW")
        break

    #let the computer play if none of the above 2 conditions happened
    computer_move()

    #check if computer won
    if(check_win("O")):
        print("COMPUTER WINS")
        break

    #check if draw happened
    if(board_full()):
        print("DRAW")
        break


```

```python
#week 2-8 puzzle using DFS:
import java.util.*;

public class USN_Week2_DFS {

    static int[] goal = {1, 2, 3, 4, 5, 6, 7, 8, 0};
    static int[] rowMove = {-1, 1, 0, 0};
    static int[] colMove = {0, 0, -1, 1};
    static String[] moveName = {"Up", "Down", "Left", "Right"};

    static boolean dfs(int[] state, Set<String> visited, List<String> path) {

        // Goal check
        if (Arrays.equals(state, goal)) {
            return true;
        }

        visited.add(Arrays.toString(state));

        // Find blank tile
        int blank = 0;
        for (int i = 0; i < 9; i++) {
            if (state[i] == 0) {
                blank = i;
                break;
            }
        }

        int row = blank / 3;
        int col = blank % 3;

        // Try all four moves
        for (int i = 0; i < 4; i++) {

            int newRow = row + rowMove[i];
            int newCol = col + colMove[i];

            // Check if move is valid
            if (newRow >= 0 && newRow < 3 &&
                newCol >= 0 && newCol < 3) {

                int newBlank = newRow * 3 + newCol;

                // Create new state
                int[] nextState = state.clone();

                nextState[blank] = nextState[newBlank];
                nextState[newBlank] = 0;

                String key = Arrays.toString(nextState);

                // If not visited, explore it
                if (!visited.contains(key)) {

                    path.add(moveName[i]);

                    if (dfs(nextState, visited, path)) {
                        return true;
                    }

                    // Backtracking
                    path.remove(path.size() - 1);
                }
            }
        }

        return false;
    }

    static void printBoard(int[] state) {

        for (int i = 0; i < 9; i++) {

            if (state[i] == 0)
                System.out.print("  ");
            else
                System.out.print(state[i] + " ");

            if (i % 3 == 2)
                System.out.println();
        }

        System.out.println();
    }

    public static void main(String[] args) {

        int[] initial = {
            1, 2, 3,
            4, 0, 6,
            7, 5, 8
        };

        Set<String> visited = new HashSet<>();
        List<String> path = new ArrayList<>();

        System.out.println("Initial State:");
        printBoard(initial);

        boolean found = dfs(initial, visited, path);

        if (found) {

            System.out.println("Goal State:");
            printBoard(goal);

            System.out.println("Solution found using DFS!");
            System.out.println("Number of moves: " + path.size());

            System.out.println("Moves:");

            for (String move : path) {
                System.out.println(move);
            }

        } else {
            System.out.println("No solution found.");
        }
    }
}
```

```python
#week 2-8 puzzle using IDS:

import java.util.*;

public class USN_Week2_IDS {

    static int[] goal = {1, 2, 3, 4, 5, 6, 7, 8, 0};

    static int[] rowMove = {-1, 1, 0, 0};
    static int[] colMove = {0, 0, -1, 1};
    static String[] moveName = {"Up", "Down", "Left", "Right"};

    static boolean depthLimitedDFS(
            int[] state,
            int depth,
            Set<String> visited,
            List<String> path) {

        // Goal check
        if (Arrays.equals(state, goal)) {
            return true;
        }

        // Depth limit reached
        if (depth == 0) {
            return false;
        }

        visited.add(Arrays.toString(state));

        // Find blank tile
        int blank = 0;

        for (int i = 0; i < 9; i++) {
            if (state[i] == 0) {
                blank = i;
                break;
            }
        }

        int row = blank / 3;
        int col = blank % 3;

        // Try all four possible moves
        for (int i = 0; i < 4; i++) {

            int newRow = row + rowMove[i];
            int newCol = col + colMove[i];

            // Check valid move
            if (newRow >= 0 && newRow < 3 &&
                newCol >= 0 && newCol < 3) {

                int newBlank = newRow * 3 + newCol;

                // Create next state
                int[] nextState = state.clone();

                nextState[blank] = nextState[newBlank];
                nextState[newBlank] = 0;

                String key = Arrays.toString(nextState);

                // Explore if not visited
                if (!visited.contains(key)) {

                    path.add(moveName[i]);

                    if (depthLimitedDFS(
                            nextState,
                            depth - 1,
                            visited,
                            path)) {

                        return true;
                    }

                    // Backtracking
                    path.remove(path.size() - 1);
                }
            }
        }

        return false;
    }

    static boolean IDS(int[] initial, List<String> path) {

        int depth = 0;

        while (true) {

            Set<String> visited = new HashSet<>();

            path.clear();

            System.out.println("Searching with depth limit: " + depth);

            if (depthLimitedDFS(
                    initial,
                    depth,
                    visited,
                    path)) {

                return true;
            }

            depth++;
        }
    }

    static void printBoard(int[] state) {

        for (int i = 0; i < 9; i++) {

            if (state[i] == 0)
                System.out.print("  ");
            else
                System.out.print(state[i] + " ");

            if (i % 3 == 2)
                System.out.println();
        }

        System.out.println();
    }

    public static void main(String[] args) {

        int[] initial = {
            1, 2, 3,
            4, 0, 6,
            7, 5, 8
        };

        List<String> path = new ArrayList<>();

        System.out.println("Initial State:");
        printBoard(initial);

        boolean found = IDS(initial, path);

        if (found) {

            System.out.println("Goal State:");
            printBoard(goal);

            System.out.println("Solution found using IDS!");
            System.out.println("Number of moves: " + path.size());

            System.out.println("Moves:");

            for (String move : path) {
                System.out.println(move);
            }

        } else {
            System.out.println("No solution found.");
        }
    }
}
```
