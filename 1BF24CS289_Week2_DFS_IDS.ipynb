```python
#WEEK 1:TIC TAC TOE IMPLEMENTATION
import random

board=[" "]*9
def print_board():
    print(board[0],"|",board[1],"|",board[2])
    print()
    print(board[3],"|",board[4],"|",board[5])
    print()
    print(board[6],"|",board[7],"|",board[8])
    print()

def board_full():
  return " " not in board

winning_combinations=[
    #first three are rows,next three columns last 2 are diagonals
    [0,1,2],[3,4,5],[6,7,8],[0,3,6],[1,4,7],[2,5,8],[0,4,8],[2,4,6]
]

def human_move():
    while(True):
        position=int(input("enter position(1 to 9):"))

        #position invalid check
        if(position<1 or position>9):
          print("enter valid position")
        #position full  check
        elif(board[position-1]!=" "):
          print("position already filled")
        #correct position entered
        else:
          board[position-1]="X"
          break

def computer_move():
    while(True):
        position=random.randint(0,8)
        if(board[position]==" "):
            board[position]="O"
            break

def check_win(Player):
    for combination in winning_combinations:
        if board[combination[0]]==board[combination[1]]==board[combination[2]]==Player:
            return True
    return False

#main code(the helper functions will be above this)
print("TIC TAC TOE game\n")
print("HUMAN:X\n")
print("COMPUTER:O\n")

while(True):
    #display the board before letting human play
    print_board()

    #human plays
    human_move()

    #display the board after human has played
    print_board()

    #check if human won
    if(check_win("X")):
        print("HUMAN WINS")
        break

    #check if draw happened
    if(board_full()):
        print("DRAW")
        break

    #let the computer play if none of the above 2 conditions happened
    computer_move()

    #check if computer won
    if(check_win("O")):
        print("COMPUTER WINS")
        break

    #check if draw happened
    if(board_full()):
        print("DRAW")
        break


```

```python
# 8-Puzzle using Depth First Search (DFS)

GOAL = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)


def get_neighbors(state):
    neighbors = []

    pos = state.index(0)
    row, col = divmod(pos, 3)

    moves = [
        (-1, 0, "Up"),
        (1, 0, "Down"),
        (0, -1, "Left"),
        (0, 1, "Right")
    ]

    for dr, dc, move in moves:
        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_pos = new_row * 3 + new_col

            new_state = list(state)
            new_state[pos], new_state[new_pos] = \
                new_state[new_pos], new_state[pos]

            neighbors.append((tuple(new_state), move))

    return neighbors


def dfs(start):
    stack = [(start, [])]
    visited = set()

    while stack:
        state, path = stack.pop()

        if state in visited:
            continue

        visited.add(state)

        if state == GOAL:
            return path

        for next_state, move in get_neighbors(state):
            if next_state not in visited:
                stack.append((next_state, path + [move]))

    return None


def print_puzzle(state):
    for i in range(0, 9, 3):
        print(state[i:i + 3])
    print()


# Initial state
start = (1, 2, 3,
         4, 0, 6,
         7, 5, 8)

print("Initial State:")
print_puzzle(start)

solution = dfs(start)

if solution:
    print("DFS Solution:")
    print(solution)
    print("Number of moves:", len(solution))
else:
    print("No solution found.")
```

```python
# 8-Puzzle using Iterative Deepening Search (IDS)

GOAL = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)


def get_neighbors(state):
    neighbors = []

    pos = state.index(0)
    row, col = divmod(pos, 3)

    moves = [
        (-1, 0, "Up"),
        (1, 0, "Down"),
        (0, -1, "Left"),
        (0, 1, "Right")
    ]

    for dr, dc, move in moves:
        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_pos = new_row * 3 + new_col

            new_state = list(state)
            new_state[pos], new_state[new_pos] = \
                new_state[new_pos], new_state[pos]

            neighbors.append((tuple(new_state), move))

    return neighbors


def depth_limited_search(state, depth, path, visited):

    if state == GOAL:
        return path

    if depth == 0:
        return None

    for next_state, move in get_neighbors(state):

        if next_state not in visited:
            visited.add(next_state)

            result = depth_limited_search(
                next_state,
                depth - 1,
                path + [move],
                visited
            )

            if result is not None:
                return result

            visited.remove(next_state)

    return None


def ids(start, max_depth=30):

    for depth in range(max_depth + 1):

        visited = {start}

        result = depth_limited_search(
            start,
            depth,
            [],
            visited
        )

        if result is not None:
            return result

    return None


def print_puzzle(state):
    for i in range(0, 9, 3):
        print(state[i:i + 3])
    print()


# Initial state
start = (1, 2, 3,
         4, 0, 6,
         7, 5, 8)

print("Initial State:")
print_puzzle(start)

solution = ids(start)

if solution:
    print("IDS Solution:")
    print(solution)
    print("Number of moves:", len(solution))
else:
    print("No solution found.")
```
